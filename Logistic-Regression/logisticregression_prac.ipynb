```python
import pandas as pd
import numpy as np
```

```python
data=pd.read_csv("IBM HR Analytics.csv")
```

```python
#use label encoder Attrition,BusinessTravel,DailyRate,Department,EducationField,Gender,JobRole,MaritalStatus,OverTime
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
data["Attrition"]=le.fit_transform(data["Attrition"])
data["BusinessTravel"]=le.fit_transform(data["BusinessTravel"])
data["Department"]=le.fit_transform(data["Department"])
data["EducationField"]=le.fit_transform(data["EducationField"])
data["Gender"]=le.fit_transform(data["Gender"])
data["JobRole"]=le.fit_transform(data["JobRole"])
data["MaritalStatus"]=le.fit_transform(data["MaritalStatus"])
data["OverTime"]=le.fit_transform(data["OverTime"])
```

```python
data.head(5)
```

```python
data.info()
```

```python
x=data.drop(["Attrition","Over18"],axis=1)
y=data["Attrition"]
```

```python
data.isnull().sum()
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
from sklearn.linear_model import LogisticRegression
model=LogisticRegression(max_iter=1000)
model.fit(x_train,y_train)
y_pred=model.predict(x_test)
```

```python
#accuracy score and confusion matrix
from sklearn.metrics import accuracy_score,confusion_matrix
accuracy_score(y_test,y_pred)
confusion_matrix(y_test,y_pred)
```

```python
from sklearn.metrics import accuracy_score,precision_score,recall_score,f1_score
from sklearn.metrics import classification_report
accuracy=accuracy_score(y_test,y_pred)
precision=precision_score(y_test,y_pred)
recall=recall_score(y_test,y_pred)
f1=f1_score(y_test,y_pred)
print("accuracy:",accuracy)
print("precision:",precision)
print("recall:",recall)
print("f1-score:",f1)
print("Classification Report:")
print(classification_report(y_test,y_pred))
```

```python
# Save model using pickle
import pickle

# Save the trained model
with open('logistic_model.pkl', 'wb') as file:
    pickle.dump(model, file)

# Save feature names
feature_names = x.columns.tolist()
with open('feature_names.pkl', 'wb') as file:
    pickle.dump(feature_names, file)

# Recreate encoders for categorical columns (since we overwrote le multiple times)
# We need to reencode the data to extract the mappings
categorical_features = ['BusinessTravel', 'Department', 'EducationField', 'Gender', 'JobRole', 'MaritalStatus', 'OverTime']
encoders = {}

# Create fresh encoders by reading the CSV again
data_fresh = pd.read_csv("IBM HR Analytics.csv")
for col in categorical_features:
    le_temp = LabelEncoder()
    le_temp.fit(data_fresh[col])
    encoders[col] = le_temp

# Save all encoders
with open('encoders.pkl', 'wb') as file:
    pickle.dump(encoders, file)

print("Model, feature names, and encoders saved successfully!")
```
